# Flood-It Environment Explorer

Simple interactive notebook to explore the Flood-It Gymnasium environment.

## Usage:
- Call `reset()` to start a new game
- Call `step(color)` to flood with a color (0-5)
- Check `env.board`, `env.moves_used`, `env.solved` for current state

```python
from floodit_env import FloodItEnv

# Create environment (no rendering)
env = FloodItEnv()

print(f"Observation space: {env.observation_space}")
print(f"Action space: {env.action_space}")
print(f"Board size: {env.observation_space.shape}")
```

```python
# Reset the environment to get a fresh game
observation, info = env.reset()

print("=== Initial Board ===")
print(observation)
print(f"\nInfo: {info}")
```

```python
# Get current board state
board = env.board
print("Current board:")
print(board)

print(f"\nMoves used: {env.moves_used}")
print(f"Solved: {env.solved}")
```

```python
# Take a step - flood with a color (0-5)
# Replace 3 below with your chosen color to play
action = 3  # <-- choose a color from 0-5 here

observation, reward, terminated, truncated, info = env.step(action)

print("=== Board After Step ===")
print(observation)
print(f"\nReward: {reward:.2f}")
print(f"Info: moves={info['moves']}, moves_left={info['moves_left']}, solved={info['solved']}")
```

```python
# Take another step
action = 1  # <-- choose a color from 0-5 here

observation, reward, terminated, truncated, info = env.step(action)

print("=== Board After Step ===")
print(observation)
print(f"\nReward: {reward:.2f}")
print(f"Info: moves={info['moves']}, moves_left={info['moves_left']}, solved={info['solved']}")
```

```python
# Close environment when done
env.close()
```
